Riazi and Daubert (1987) developed a simple two-parameter equation for predicting the physical properties of pure compounds and undefined hydrocarbon mixtures.


Based on the use of the molecular weight M and specific gravity  of the undefined petroleum fraction as the correlating parameters.

```python
import math
#Estimate the critical properties of the heptanes-plus fraction, i.e., C7+,

M=float(input("molecular weight: "))
g=float(input("specific gravity: "))

def crit_temp(M,g):
  a=544.2
  b=0.2998
  c=1.0555
  d=-1.3478*(10**-4)
  e=-0.61641
  f=0
  Tc=a * (M**b) * (g**c) * math.exp(d*M + e*g + f*M*g)
  return Tc

def crit_pressure(M,g):
  a=4.5203*(10**4)
  b=-0.8063
  c=1.6015
  d=-1.8078*(10**-3)
  e=-0.3084
  f=0
  Pc=a * (M**b) * (g**c) * math.exp(d*M + e*g + f*M*g)
  return Pc

def crit_volume(M,g):
  a=1.206*(10**-2)
  b=0.20378
  c=-1.3036
  d=-2.657*(10**-3)
  e=0.5287
  f=2.6012*(10**-3)
  Vc=a * (M**b) * (g**c) * math.exp(d*M + e*g + f*M*g)
  return Vc

def boiling_p_temp(M,g):
  a=6.77857
  b=0.401673
  c=-1.58262
  d=3.77409*(10**-3)
  e=2.984036
  f=-4.25288*(10**-3)
  Tb=a * (M**b) * (g**c) * math.exp(d*M + e*g + f*M*g)
  return Tb


print(f"critical temperature: {crit_temp(M,g):.1f} R")
print(f"critical pressure: {crit_pressure(M,g):.1f} Psia")
print(f"critical volume: {crit_volume(M,g):.5f} ft3/lb")
print(f"boiling point temperature: {boiling_p_temp(M,g):.2f} R")
```

Edmister (1958) proposed a correlation for estimating the acentric factor T of pure fluids and petroleum fractions. The equation, widely used in the petroleum industry, requires boiling point, critical temperature, and critical pressure.

```python
Tc= round(crit_temp(M,g), 1)
Tb=round(boiling_p_temp(M,g), 2)
Pc=round(crit_pressure(M,g), 1)
Vc=round(crit_volume(M,g), 5)

#Estimate the acentric factor using Edmister’s Equation
def acentric_factor(Tc,Tb,Pc):
  y=(math.log10(Pc/14.7))*3
  x= Tc/Tb
  z= 7*(x-1)
  omega= (y/z)-1
  return omega

print(f"{acentric_factor(Tc,Tb,Pc):.4f}")
```

The critical compressibility factor is another property that is often used in thermodynamic-property prediction models. It is defined as the component compressibility factor calculated at its critical point.

```python
def crit_compressibility_factor(Tc,M,Pc,Vc):
  R=10.73
  Zc= (Pc*Vc*M)/(R*Tc)
  return Zc

#Other Method for Critical Compressibility Estimation
omega=round(acentric_factor(Tc,Tb,Pc), 4)

def crit_compressibility_factor_Haugen(omega):
  Zch=1/(1.28*omega+3.41)
  return Zch
def crit_compressibility_factor_Reid(omega):
  Zcr=0.291-0.080*omega
  return Zcr
def crit_compressibility_factor_Salerno(omega):
  Zcs=0.291-0.080*omega-0.016*omega**2
  return Zcs
crit_compressibility_factor_Nath=0.2918-0.0928


print(omega)
print(f"{crit_compressibility_factor_Haugen(omega):.4f}")
print(f"{crit_compressibility_factor_Reid(omega):.4f}")
print(f"{crit_compressibility_factor_Salerno(omega):.4f}")
print(f"{crit_compressibility_factor_Nath:.4f}")

```
