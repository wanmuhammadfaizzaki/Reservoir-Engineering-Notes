```python
Res_Temp = input("Your reservoir temperature > critical temperature (C)? (yes/no):").lower()
if Res_Temp == 'yes':
    result="Gas reservoir"
else:
    result="Oil reservoir"
print(result)
```

```python
have_bo_input = input("Do you have a value for oil formation volume factor (Bo)? (yes/no):").lower()
if have_bo_input == 'yes':
    Bo = float(input("Enter the value for oil formation value factor in bbl/STB:"))
else:
    Bo = "NA" # Default value if no Bo is provided
    print("Bo value is not available")

GOR = float(input("Enter the value for Gas-oil ratio in scf/STB:"))

have_oil_gravity_input = input("Do you have a value for oil gravity? (yes/no):").lower()
if have_oil_gravity_input == 'yes':
    Oil_gravity = float(input("Enter the value for oil gravity in API:"))
else:
    Oil_gravity = "NA" # Default value if no GOR is provided
    print("GOR value is not available")
```

```python
if Res_Temp == 'yes': # This indicates a Gas reservoir
    # Gas reservoir classification
    if GOR >= 8000 and GOR <= 70000 and Oil_gravity > 50:
      result="Retrograde gas-condensate reservoir"
    elif GOR >= 60000 and GOR <= 100000 and Oil_gravity > 60:
      result="Wet-gas reservoir"
    elif GOR > 100000:
      result="Dry-gas reservoir"
    else:
      result="Cannot define (Gas reservoir parameters do not match)"
else: # This indicates an Oil reservoir
    # Oil reservoir classification
    if Bo == "NA" and GOR >= 200 and GOR <= 700 and Oil_gravity >= 15 and Oil_gravity <= 40:
      result="Ordinary black oil"
    elif Bo != "NA" and Bo < 1.2 and GOR < 200 and Oil_gravity < 35:
      result="Low-shrinkage oil"
    elif Bo != "NA" and Bo < 2 and GOR >= 2000 and GOR <= 3200 and Oil_gravity >= 45 and Oil_gravity <= 55:
      result="Volatile crude oil or high shrinkage crude oil"
    elif Bo != "NA" and Bo > 2 and GOR > 3000:
      result="Near-critical crude oil"
    else:
      result="Cannot define (Oil reservoir parameters do not match)"

print(f"The reservoir type is: {result}")
```

```python
print(f"The reservoir type is: {result}")
print('The characteristic is :')
if result == "Retrograde gas-condensate reservoir":
  print("- Reservoir temperature lies between the critical temperature (Tc) and cricondentherm (Tct)")
  print("- Water-white and slightly colored")
elif result == "Wet-gas reservoir":
  print("- Reservoir temperature is above cricondentherm (Tct)")
  print("- Separator con. lie within the two phase region")
elif result == "Dry-gas reservoir":
  print("- Reservoir temperature is above cricondentherm (Tct)")
  print("- HC exist as gas both in the reservoir and in the surface facilities")
  print("- Only liquid associated with gas is water")
elif result == "Ordinary black oil":
  print("- Quality lines which are approximately equally spaced characterize this black oil phase diagram")
  print("- The liquid shrinkage curve approximates a straight line except at very low pressures")
elif result == "Low-shrinkage oil":
  print("- Quality lines that are closely spaced near the dew-point curve")
  print("- Black or deeply colored")
  print("- Substantial liquid recovery at separator conditions is around 85%")
elif result == "Volatile crude oil or high shrinkage crude oil":
  print("- Quality lines are close together near the bubble-point and are more widely spaced at lower pressures")
  print("- Lower liquid recovery of separator conditions")
  print("- Greenish to orange in color")
  print("- API gravity of the stock-tank liquid will increase in the later life of the reservoirs")
elif result == "Near-critical crude oil":
  print("- Reservoir temperature T is near the critical temperature Tc of the hydrocarbon system")
  print("- Crude oil may shrink from 100% to 55% or less at 10 to 50 psi below bubble point")
  print("- Heptanes-plus(12.5-20 mol%), 35% or more of ethane through hexanes, and the remainder methane")
elif result == "Cannot define (Gas reservoir parameters do not match)":
  print("Cannot define")
elif result == "Cannot define (Oil reservoir parameters do not match)":
  print("Cannot define")
else:
  print("Cannot define")
```
